```python
import pandas as pd
import numpy as np
from pandas import Series,DataFrame
#python数据可视化
import matplotlib.pyplot as plt
plt.rcParams['font.family'] = ['SimHei']
```

```python
#绘制线图
def fun1():
    x = np.arange(0,10,2)
    y = 2*x
    #获取画板
    plt.figure()
    #绘制
    plt.plot(x,y,'--o',color='r')
fun1()
```

```python
#绘制二次函数
def fun2():
    x = np.arange(-5,5,1)
    y = x ** 2 + 1
    plt.figure()
    plt.plot(x,y,'--x',color = 'b')
fun2()
```

```python
#绘制三角函数
def fun3():
    x = np.linspace(1,10,100)
    y = np.cos(x)
    z = np.sin(x)
    plt.figure()
    plt.plot(x,y,'-2')
    plt.plot(x,z,'-2')
fun3()
```

```python
def fun4():
    x = np.arange(0,10,1)
    y = x ** 2 + 1
    #设置画板尺寸（inche）宽和高
    plt.figure(figsize = (8,4))
    plt.plot(x,y,'-x',color = 'b')
    
    #设置xy轴注释
    plt.xlabel('x轴')
    plt.ylabel('y轴')
    
    #修改刻度
    plt.xticks(np.arange(0,10))
    plt.yticks(y)

    #在图中添加注释
    #先获取坐标轴
    axis = plt.gca()#gca = get current axis
    for i,j in zip(x,y):
        axis.text(i,j + 0.5,j, bbox=dict(facecolor='red', alpha=0.5))

    #在当前画板中添加表格背景
    plt.grid(True)
fun4()
```
