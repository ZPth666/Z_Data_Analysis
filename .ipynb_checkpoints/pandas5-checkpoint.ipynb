```python
#二手房数据分析
import numpy as np
import pandas as pd
from pandas import DataFrame,Series
```

```python
path = 'fangzi.csv'
data = pd.read_csv(path)
data
```

```python
data.shape
```

```python
data.columns
```

```python
#建造最早的房子的成交价格
#先查看数据集中是否含有脏数据
s = data['Construction time']
s
```

```python
s_num = pd.to_numeric(s,errors = 'coerce')
s_num
```

```python
s_num[s_num < 1000].value_counts()
```

```python
index1 = s_num[s_num.isnull()].index
index2 = s_num[(s_num == 0)|(s_num == 1)].index
index2
```

```python
#根据索引删除数据集中的数据：.drop(index,inplace = )
data.drop(index1, inplace = True)
data
```

```python
data.drop(index2, inplace = True)
data
```

```python
#检查脏数据是否被删除
data['Construction time'].value_counts()
```

```python
data.sort_values(['Construction time','Total price'])[['Construction time','Total price']]
```

```python
#哪个区域卖出的房子最多？
data.groupby(['District']).count()['URL']
```

```python
#不分组做法
data.District.value_counts()
```

```python
#哪个区的房子平均价格最高
data.groupby(['District'])['Total price'].mean().sort_values(ascending = False)
```

```python
#海淀区的平均房价(每平米)
data[data['District']=='海淀']['Price'].mean()
```

```python
#西城区的房子中有多少两居室
data[(data.District=='西城') & (data['Bed Room']==2)].count()
```
