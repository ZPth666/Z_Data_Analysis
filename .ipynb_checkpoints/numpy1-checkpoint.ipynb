```python
import numpy as np
# 使用时先引用
```

```python
arr = np.array([1,2,3,4])
arr
# 创建n维数组 np.array()
```

```python
arr.ndim
# np.array的属性ndim表示数组的维度
```

```python
arr.shape
# np.array的属性ndim表示数组的结构（长宽高）
```

```python
arr.size
# np.array的属性zize表示数组中的个数
```

```python
#np.zeros(shape):返回一个n维0填充的数组
arr1 = np.zeros((2,3,4))
```

```python
arr1
```

```python
#np.ones(shape):返回一个n维1填充的数组
arr2 = np.ones((2,3,4))
arr2
```

```python
#np.zeros_like(数组) np.ones_like(数组)
#对数组进行0/1化，生成新数组
a = np.zeros_like(arr1)
a
```

```python
#np.full(结构，值):将某个结构的数组全部赋同样值
b = np.full((2,2,3),6)
b
```

```python
b = np.full((2,2,3),np.pi)
b
```

```python
#np.linspace(start,end,num = n)
arr3 = np.linspace(1,6,num = 5,endpoint = True)
arr3
```

```python
#?np.function:查函数参数
?np.linspace
```

```python
#np.fromfunction(lambda x,y,...:...,(nx,ny,...))
#由lambda表达式创建出n维数组
m = np.fromfunction(lambda x,y,z:(x+1)*(y+1)*(z+1),(6,6,6))
```

```python
m
```

```python
#创建n维数组时，可以指定数据类型，加入dtype
ar = np.array([1,2,3,4,5],dtype = np.float64)
ar
```

```python
?np.array
```

```python
#创建n维数组时，若对象类型不统一，会自动转化为更大的数据类型
arr = np.array([1,2,3,True,'a',False])
arr
```

```python
#产生随机数
#根据指定范围产生指定结构的随机数数组
r = np.random.randint(0,100,(2,3,4)) #整数数组
r1 = np.random.randn(2,3,4) #符合正态分布,均值为0，方差为1
r2 = np.random.uniform(0,100,(2,3,4)) #随机数
```

```python
?np.random.randn
```
