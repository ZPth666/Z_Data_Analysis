```python
import pandas as pd
import numpy as np
from pandas import Series,DataFrame
```

```python
#导入文件利用路径变量导入，分为相对引用和绝对引用
path = "Chipotle_datas.tsv"
df = pd.read_csv(path, sep = "\t")
#利用pd.read_csv转化为DataFrame，参数中写入路径和分隔符（默认用逗号）
df
```

```python
#查看数据集的结构：shape,columns,info
```

```python
df.tail(7)
```

```python
#统计数据集中某一列中每一个数据重复出现的次数：.value_counts()   （得到一个series）
df['item_name'].value_counts(ascending = True)
```

```python
#统计数据集中出现数据的种类
df['item_name'].value_counts(ascending = True).shape
```

```python
#还可以用.nunique
df['item_name'].nunique()
```

```python
#统计choice_description中出现最多的选项
df['choice_description'].value_counts().head(1)
```

```python
df['quantity'].sum()
```

```python
#对某一列进行算法的操作与处理: .apply() 参数为函数或者lambda表达式
df['item_price'].apply(lambda x: float(x[1:]))
```

```python
#创建新的一列，直接写新索引然后赋值
df['item_price_float'] = df['item_price'].apply(lambda x: float(x[1:]))
df.head()
```

```python
#计算共有多少单订单
df.order_id.nunique()
```

```python
df.order_id.value_counts().count()
```

```python
df.order_id.value_counts().shape
```

```python
#分别计算每一单的总价，然后求均值（分组求和+求均值）
#分组函数：.groupby()
#求平均值函数：.mean()[]
df.groupby(['order_id']).sum()['item_price_float'].mean()
```
