```python
#数据的合并
import numpy as np
import pandas as pd
from pandas import DataFrame, Series
```

```python
arr = np.random.randint(0,10,(3,3))
arr
```

```python
#numpy中的concatenate支持多维数组数据合并
arr1 = np.concatenate((arr,arr),axis = 0)
arr1
```

```python
arr2 = np.concatenate((arr,arr),axis = 1)
arr2
```

```python
#利用列表表达式创建字典，再利用字典创建DataFrame
dict = {i:[p*i for p in range(3)] for i in range(10)}
dict
```

```python
def fun1(cols,index):
    data = {i:[str(p)+i for p in index] for i in cols}
    return pd.DataFrame(data,columns = cols,index = index)
fun1(['a','b'],['1','2'])
```

```python
#pandas中的concat可以实现基本的DataFrame的间数据合并
#pd.concat(objs, axis=0, join='outer', join_axes=None, ignore_index=False, keys=None, levels=None, names=None, sort=None,verify_integrity=False, sort=None, copy=True)
df1 = fun1(["a","b"],[1,2])
df2 = fun1(["a","b"],[3,4])
display(df1,df2)
```

```python
pd.concat((df1,df2),axis = 0) #axis参数可修改合并的方向
```

```python
#合并时索引可以重复
df3 = fun1(["a","b","c"],[1,2,3])
df4 = fun1(["a","b","c"],[3,4])
pd.concat((df3,df4))
```

```python
#可以在合并时重新设置索引
pd.concat((df3,df4), ignore_index = True)
```

```python
#合并时，可能出现列名不匹配的情况，利用join与join_axes参数设置数据连接的方式
#1.外连接outer（默认）相当于数据库中的全外连接（求并集）
#2.内连接inner，只显示互相匹配得上的数据（求交集）
```

```python
d1 = fun1(["A","B","C"],[1,3,5])
d2 = fun1(["a","b","c"],[2,3,4])
display(d1,d2)
```

```python
pd.concat((d1,d2),axis = 1,join = "outer")
```

```python
pd.concat((d1,d2),axis = 1,join = "inner")
```

```python
#3.左/右连接，设置合并标准（以其中一个表为标准）
d3 = fun1(["A","B","C"],[1,3,5])
d4 = fun1(["A","C","D"],[1,3,4])
df1 = pd.concat((d3,d4))
```

```python
df2 = df1[d4.columns]
df2
```

```python
df3 = df1[d3.columns]
df3
```

```python
#pandas中的merge较为常用，在进行数据合并时，一定是基于某一列进行数据合并的（默认内连接，默认根据所有列名全匹配）
#pd.merge(left ,right, how = 'inner', on = None, left_on = None, right_on = None, left_index = False, right_index = False, sort = False, suffixes = ('_x', '_y'), copy = None, indicator = False, validate = None,)
da1 = DataFrame({"employee":["Sheldon","Penny", "Kite"],"dept":["hr","marketing","sales"]})
da2 = DataFrame({"employee":["Sheldon","Penny","Bill"],"work_time":[2,4,5]})
display(da1,da2)
```

```python
pd.merge(da1,da2)
```

```python
#一对多场景
da3 = DataFrame({"employee":["Sheldon","Penny", "Penny"],"dept":["hr","marketing","sales"]})
da4 = DataFrame({"employee":["Sheldon","Penny","Bill"],"work_time":[2,4,5]})
pd.merge(da3,da4)
```

```python
#利用参数on，可确定根据哪一列参数进行合并
d1 = DataFrame({"employee":["Sheldon","Penny", "Lanard"],"dept":["hr","marketing","sales"],"salary":[10000,9000,8500]})
d2 = DataFrame({"employee":["Sheldon","Penny", "Kite"],"dept":["marketing","marketing","sales"],"work_time":[2,4,5]})
display(d1,d2)
```

```python
pd.merge(d1,d2)
```

```python
pd.merge(d1,d2,on = "employee")
```

```python
 pd.merge(d1,d2,on = "dept")
```

```python
#若左右两表格所需要匹配的列名不同，可用left_on与right_on参数进行匹配
pd.merge(d1,d2,left_on="employee", right_on="employee", how="outer")
```

```python
#可用how参数改变连接方式：内外左右
pd.merge(d1,d2,left_on="employee", right_on="employee", how="left")
```

```python
pd.merge(d1,d2,left_on="employee", right_on="employee", how="right")
```
