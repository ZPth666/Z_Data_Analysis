```python
#王者荣耀数据分析
import pandas as pd
import numpy as np
from pandas import Series,DataFrame
```

```python
path = 'type_hero.csv'
hero = pd.read_csv(path)
hero
```

```python
hero.columns
```

```python
hero.shape
```

```python
#寻找单列中的最大值（排序）：.sort_values(by=[''])
hero.sort_values(by = ['最大生命'],ascending = False)
```

```python
#查看分组后的平均值 注:求平均值若只对数字类型求，需在参数中添加numeric_only = True
#查看不同位置的英雄的数值平均值
```

```python
h = hero.groupby(by = ['英雄类型']).mean(numeric_only = True)
h.sort_values(['物理防御'],ascending = False)
```

```python
#对所有列进行函数运算：.applymap?即将废弃
#保留三位小数
h.round(3)
```

```python
h*2+6
```

```python
#遍历进行批量操作:apply和map结合使用
h.apply(lambda col: col.map(lambda x: round(x,3)))
```

```python
#查看分组后的重复数
#统计近战和远战英雄的数量
hero.groupby(['攻击范围']).count()['最大生命']
```

```python
#查看每一种英雄生命值回复平均值排名
hero.groupby(['英雄类型'])['生命回复'].mean().sort_values(ascending = False).round(3)
```
