```python
import pandas as pd
import numpy as np
from pandas import Series,DataFrame
#python数据可视化
import matplotlib.pyplot as plt
#设置中文字体
plt.rcParams['font.family'] = ['SimHei']
#显示负号
plt.rcParams['axes.unicode_minus'] = False
```

```python
#画六条不同的线图，运用正态分布的随机数
def fun1():
    x = np.arange(50)
    #列表存放颜色
    colors = ['r','b','y','b','g','c']
    #字典存放样式
    kind = {"circle":"o","start":"*","traggle_donw":"v","dash_dot":"-","dash_line":"--","x-mark":"x"}
    plt.figure(figsize = (12,8))
    for co,key in zip(colors,kind.keys()):
        y = np.random.randn(50)
        plt.plot(x,y,kind[key],color = co, label = key)
    plt.legend()
fun1()
```

```python
#标准直角坐标系下绘制三角函数
def fun2():
    x = np.linspace(-np.pi,np.pi,256)
    c = np.cos(x)
    s = np.sin(x) 
    plt.figure(figsize=(8,6))
    plt.plot(x, c, "-") 
    plt.plot(x, s, "--")
    
    #指定x轴和y轴的取值范围
    plt.xlim([-4,4]) 
    plt.ylim([-1,1])
    
    #修改刻度
    plt.xticks([-np.pi, -np.pi/2, 0, np.pi/2, np.pi],["$-\pi$","$-\pi/2$","$0$","$\pi/2$","$\pi$"]) 
    plt.yticks([-1,0,1])
    
    #去除上、右边框 
    ax = plt.gca()
    ax.spines["right"].set_color("none") 
    ax.spines["top"].set_color("none")
    #将底边框移到上面0点位置
    ax.spines["bottom"].set_position(("data",0))
    #将左边框移到右侧0点位置
    ax.spines["left"].set_position(("data",0))

    #在图上标点
    t = 2 * np.pi / 3
    y = np.sin(t)
    plt.scatter(t, y, 50, color="r")
    plt.plot((t,t),(0,y),"--", color="r")
    plt.plot((0,t),(y,y),"--", color="r")
    plt.annotate('($2\pi/3$,$\sqrt{3}/2$)',(t,y))
fun2()
```

```python
#横向柱状图 
def fun4():
    price=[15000,23000,30000,50000]#第一个参数表示y的点，第二个参数表示×的点
    plt.barh(np.arange(4),price,align="center", color="steelblue") 
    plt.xlim([0,70000])
    for x,y in zip(np.arange(4),price):
        plt.text(y,x,y)
fun4()
```

```python
#画出两个柱状图的对比图
def fun1():
    x = np.arange(4)
    y2017 = [13500,14500, 25000,40000] 
    y2018 =[16000,24000,37000,55000]
    labels = ["北京","深圳","上海","杭州"]

    #设置每根柱子的宽度
    bar_width = 0.4
    plt.bar(x, y2017, width = bar_width, color = 'steelblue', alpha = 0.8)
    plt.bar(x + bar_width, y2018, width = bar_width, color = 'indianred', alpha = 0.8)

    #改变下标
    plt.xticks(x+0.2,labels)

    #添加文字
    for x1,y1 in zip(x,y2017):
        plt.text(x1-0.15,y1+500,y1)

    for x2,y2 in zip(x,y2018):
        plt.text(x2+0.25,y2+500,y2)
fun1()
```

```python
import numpy as np
import pandas as pd
from pandas import Series, DataFrame
import matplotlib.pyplot as plt
plt.rcParams['font.family']=['SimHei']
plt.rcParams['axes.unicode_minus']=False
```

```python
#绘制饼图（使用pie）
def fun1():
    #就业率
    edu = [0.2515, 0.3724,0.3336,0.0368,0.0057]
    #学历
    labels=["中专","大专","学士","硕士","博士"]
    #饼的颜色
    colors=["#9999ff","#ff9999","#7777aa","#2442aa","#dd5555"]

    #将横、纵坐标进行标准化处理，，保证饼图是正圆的
    plt.axes(aspect="equal")
    
    plt.xlim([0,4])
    plt.ylim([0,4])
    explode = [0.1,0.1,0.1,0.2,0.2]#板块分离度
    plt.pie(x=edu, #绘图数据
            labels=labels, #数据的标签 
            colors=colors, #饼图的颜色
            autopct="%.2f%%", #设置百分比，小数点后保留2位
            startangle=180, #设置图的初始角度
            frame=1, #是否显示边框
            center=(2,2), #设置饼图原点
            explode=explode, #设置突出显示 
            radius=1.5) #设置饼的半径
fun1()
```

```python
#确定脏数据的占比：若比重小可删除
n = np.sum([info.Boarding_time == "未上牌"])
n / info.shape[0]
```

```python
#利用索引drop删除或者直接赋值
info = info[info.Boarding_time != "未上牌"]
info
```

```python
#删除后索引混乱，可重新赋值
info.index = np.arange(0,info.shape[0])
info
```

```python
#通过切片提取上牌时间中的年份和月份
info.loc[:,"year"] = info.Boarding_time.str[:4].astype("int")
info.loc[:,"month"] = info.Boarding_time.str[5:-1].astype("int")
info
```

```python
#若当前数据集是在2017年10月统计的，新增一列存放使用月份
info.loc[:,"Use_time"] = (2017-info.year)*12 + 10 - info.month
info
```

```python
#处理Km列
#将字符串转化为可处理的浮点数据，产生报错————有脏数据待处理
#通过索引加切片得到干净数据（利用contains函数）
info = info[~info.Km.str.contains("百")]
info.loc[:,"New_Km"] = info.Km.str[:-3].astype("float")
info
```

```python
#处理New_price列数据
#转换过程仍发现脏数据需要处理
info = info[~info.New_price.str.contains("暂")]
info.loc[:,"New_New_price"] = info.New_price.str[:-1].astype("float")
info
```

```python
#查看不同价格区间的二手车数量及占比
info.Sec_price.value_counts()
```

```python
#找出区间上限与下限
info.Sec_price.sort_values()
price_max = 1000
price_min = 0
price_bins = [price_min, 3, 5, 8, 10, 15, 20, 30, 50, 100, price_max]
```

```python
#利用pandas的cut函数完成对区间的划分
price_cut = pd.cut(info.Sec_price, bins = price_bins)
price_result = price_cut.value_counts().sort_index()
price_result
```

```python
#画柱状图以及饼状图
def fun1():
    x = np.arange(len(price_result))
    y = price_result
    plt.figure(figsize = (8,6))
    plt.bar(x,y,color = "steelblue",alpha = 0.8)
    plt.xlabel("价格区间")
    plt.ylabel("数量")
    plt.title("二手车价格分布")
    plt.xticks(np.arange(len(price_result)), price_result.index, rotation = 30)
    plt.ylim([0,2200])

    for a,b,c in zip(x-0.3,y+10,price_percent):
        plt.text(a,b,c)
fun1()
```

```python
def fun2():
    L = [i for i in price_result / price_result.sum()]#占比情况

    labels = price_result.index#区间
    
    colors=["#9999ff","#ff9999","#7777aa","#2442aa","#dd5555","#0000ff","#00ff00","#ff0000","#ffff00","#00ffff"]#颜色
    
    explode = [0.1,0.1,0.1,0.2,0.2,0.1,0.2,0.2,0.15,0.1]#板块分离度
    
    plt.pie(x=L, #绘图数据
        labels=labels, #数据标签 
        colors=colors, #饼图颜色
        autopct="%.2f%%", #设置百分比，小数点后保留2位
        startangle=300, #设置图的初始角度
        pctdistance = 0.6, #百分比数字距离圆心的远近
        labeldistance = 1.15, #标签距离数字的远近
        frame=0, #是否显示边框
        center=(2,2), #设置饼图原点
        explode=explode, #设置突出显示
        radius=1, #设置饼的半径
        textprops = {"fontsize":10}, #全局字体大小设置
        wedgeprops = {"edgecolor":"black", "linewidth":1}) #扇区间分割线
    
    plt.title("二手车价格分布(单位：万)",fontsize = 12,pad = 12)
fun2()
```

```python
#利用列表表达式创建字典，再利用字典创建DataFrame
dict = {i:[p*i for p in range(3)] for i in range(10)}
dict
```

```python
def fun1(cols,index):
    data = {i:[str(p)+i for p in index] for i in cols}
    return pd.DataFrame(data,columns = cols,index = index)
fun1(['a','b'],['1','2'])
```

```python
#pandas中的concat可以实现基本的DataFrame的间数据合并
#pd.concat(objs, axis=0, join='outer', join_axes=None, ignore_index=False, keys=None, levels=None, names=None, sort=None,verify_integrity=False, sort=None, copy=True)
df1 = fun1(["a","b"],[1,2])
df2 = fun1(["a","b"],[3,4])
display(df1,df2)
```

```python
pd.concat((df1,df2),axis = 0) #axis参数可修改合并的方向
```

```python
#合并时索引可以重复
df3 = fun1(["a","b","c"],[1,2,3])
df4 = fun1(["a","b","c"],[3,4])
pd.concat((df3,df4))
```

```python
#可以在合并时重新设置索引
pd.concat((df3,df4), ignore_index = True)
```

```python
#合并时，可能出现列名不匹配的情况，利用join与join_axes参数设置数据连接的方式
#1.外连接outer（默认）相当于数据库中的全外连接（求并集）
#2.内连接inner，只显示互相匹配得上的数据（求交集）
```

```python
d1 = fun1(["A","B","C"],[1,3,5])
d2 = fun1(["a","b","c"],[2,3,4])
display(d1,d2)
```

```python
pd.concat((d1,d2),axis = 1,join = "outer")
```

```python
pd.concat((d1,d2),axis = 1,join = "inner")
```

```python
#3.左/右连接，设置合并标准（以其中一个表为标准）
d3 = fun1(["A","B","C"],[1,3,5])
d4 = fun1(["A","C","D"],[1,3,4])
df1 = pd.concat((d3,d4))
```

```python
#一对多场景
da3 = DataFrame({"employee":["Sheldon","Penny", "Penny"],"dept":["hr","marketing","sales"]})
da4 = DataFrame({"employee":["Sheldon","Penny","Bill"],"work_time":[2,4,5]})
pd.merge(da3,da4)
```

```python
#利用参数on，可确定根据哪一列参数进行合并
d1 = DataFrame({"employee":["Sheldon","Penny", "Lanard"],"dept":["hr","marketing","sales"],"salary":[10000,9000,8500]})
d2 = DataFrame({"employee":["Sheldon","Penny", "Kite"],"dept":["marketing","marketing","sales"],"work_time":[2,4,5]})
display(d1,d2)
```

```python
pd.merge(d1,d2)
```

```python
pd.merge(d1,d2,on = "employee")
```

```python
 pd.merge(d1,d2,on = "dept")
```

```python
#若左右两表格所需要匹配的列名不同，可用left_on与right_on参数进行匹配
pd.merge(d1,d2,left_on="employee", right_on="employee", how="outer")
```

```python
#可用how参数改变连接方式：内外左右
pd.merge(d1,d2,left_on="employee", right_on="employee", how="left")
```

```python
pd.merge(d1,d2,left_on="employee", right_on="employee", how="right")
```
