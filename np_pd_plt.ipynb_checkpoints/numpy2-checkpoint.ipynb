```python
#布尔索引:根据条件可得到布尔数组，将布尔数组放入下标即可筛选出满足条件的元素
#可通过布尔索引访问和修改
arr2 = np.random.rand(5,6)
arr2
```

```python
arr2 > 0.5
```

```python
arr2[arr2>0.5]
```

```python
arr2[arr2<0.5] = 0.5
arr2
```

```python
(arr2>0.8) | (arr2<0.6)
```

```python
arr2[(arr2>0.8) | (arr2<0.6)] = 0.7
arr2
```
