```python
import pandas as pd
import numpy as np
from pandas import Series,DataFrame
```

```python
#Series可以看作是带下标的一维数组（数字索引默认为数字，可以加入index参数指定索引）
s = Series(['a','b','c','d','e'],index = [1,2,4,6,8])
s
```

```python
#Series有了values和index两个属性
s.values
```

```python
s.index
```

```python
#Series支持索引和切片
#可以通过多个索引取值
#可以用布尔索引取值
#可以对series中的值进行运算
```

```python
#Series可以通过字典创建，key -> values(若有index参数，则覆盖)
data = {'name':'zhao', 'age':18, 'height':180}
s1 = Series(data)
s1
```

```python
s2 = Series(data, index = ['name','age','weight'])
s2 #NaN表示为空值（Not a Number）
```

```python
#pd.isnull(s):相当于s.values == NaN（布尔索引）
#可用于计算时将空值赋值为0
pd.isnull(s2)
```

```python
s2[pd.isnull(s2)]
```

```python
#pd.notnull获取不为空的
```

```python
#Series可以直接相加
#若索引相同则相加，不同则保留并赋值为NaN
```

```python
s1 + s2
```

```python
#DataFrame 是二维表
data = {'city':['北京','上海','广州','深圳','杭州'],
        'year':[2022,2024,2025,2025,2026],
        'people':[1500,2500,2000,2500,1500]}
df = DataFrame(data)
df
```

```python
#DateFrame的三个属性：columns,info,shape
df.columns
```

```python
df.info
```

```python
df.shape
```

```python
#在创建df时，可自主指定index和columns参数
#若有不匹配的columns参数，则赋值为NaN
```

```python
df1 = DataFrame(data,
                   columns = ['city', 'year', 'people','dept'],
                   index = ['4','5','6','7','8'])
df1
```

```python
#通过多个索引可以获取DataFrame中的多个值以及赋值
#行列索引，切片 
df1[['city','year']]
```

```python
df1['dept'] = np.arange(5)
df1
```

```python
#行索引：df.loc[]
df1.loc[['4','5']]
```

```python
#行列同时索引，仍然用loc
df1.loc[['4','5'],['city','year']]
```

```python
#按照行列次序索引，可用iloc切片
df1.iloc[1:3,2:4]
```
