```python
import numpy as np
import pandas as pd
from pandas import Series, DataFrame
import matplotlib.pyplot as plt
plt.rcParams['font.family']=['SimHei']
plt.rcParams['axes.unicode_minus']=False
```

```python
#绘制饼图（使用pie）
def fun1():
    #就业率
    edu = [0.2515, 0.3724,0.3336,0.0368,0.0057]
    #学历
    labels=["中专","大专","学士","硕士","博士"]
    #饼的颜色
    colors=["#9999ff","#ff9999","#7777aa","#2442aa","#dd5555"]

    #将横、纵坐标进行标准化处理，，保证饼图是正圆的
    plt.axes(aspect="equal")
    
    plt.xlim([0,4])
    plt.ylim([0,4])
    explode = [0.1,0.1,0.1,0.2,0.2]#板块分离度
    plt.pie(x=edu, #绘图数据
            labels=labels, #数据的标签 
            colors=colors, #饼图的颜色
            autopct="%.2f%%", #设置百分比，小数点后保留2位
            startangle=180, #设置图的初始角度
            frame=1, #是否显示边框
            center=(2,2), #设置饼图原点
            explode=explode, #设置突出显示 
            radius=1.5) #设置饼的半径
fun1()
```

```python
#直方图，类似柱状图，用来统计数据出现的频率/概率（hist）
def fun2():
    x = np.random.randint(0,10,20)
    print(x)
    plt.hist(x,orientation = 'vertical')#horizontal
fun2()
```

```python
#散点图（scatter）
def fun3():
    x = np.random.randn(1000)
    y = np.random.randn(1000)
    colors = np.random.random(3000).reshape((1000,3))
    plt.scatter(x,y,color = colors,s = 10)
fun3()
```

```python
#子图，可以在一个画布中画多张图
def fun4():
    plt.figure(figsize=(12,8))
    
    a1 = plt.subplot(2,3,1)
    x1 = np.arange(5) 
    y1 = x1 + 5
    a1.plot(x1,y1)
    a1.grid(color="r",linestyle="--") 
    
    a2 = plt.subplot(2,3,2)
    x2 = np.linspace(-np.pi,np.pi,256)
    y2 = np.sin(x2) 
    a2.plot(x2,y2)
    a2.grid(color="g",linestyle="-.") 
    
    a3 = plt.subplot(2,3,3)
    x3 = np.linspace(-np.pi,np.pi,256)
    y3 = np.cos(x3) 
    a3.plot(x3,y3)
    a3.grid(color="b",linestyle=":")

    a4=plt.subplot(2,3,4)
    x4 = np.arange(5) 
    y4 = x4 + 5
    a4.plot(x4,y4)
    a4.grid(color="r",linestyle="--")
    
    a5 = plt.subplot(2,3,5)
    x5 = np.linspace(-np.pi,np.pi,256)
    y5 = np.sin(x2)
    a5.plot(x5,y5)
    a5.grid(color="g",linestyle="-.") 
    
    a6 = plt.subplot(2,3,6)
    x6 = np.linspace(-np.pi,np.pi,256)
    y6=np.cos(x6) 
    a6.plot(x6,y6)
    a6.grid(color="b",linestyle=":")
fun4()
```
