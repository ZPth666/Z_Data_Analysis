```python
import pandas as pd
import numpy as np
from pandas import Series,DataFrame
#python数据可视化
import matplotlib.pyplot as plt
#设置中文字体
plt.rcParams['font.family'] = ['SimHei']
#显示负号
plt.rcParams['axes.unicode_minus'] = False
```

```python
#画六条不同的线图，运用正态分布的随机数
def fun1():
    x = np.arange(50)
    #列表存放颜色
    colors = ['r','b','y','b','g','c']
    #字典存放样式
    kind = {"circle":"o","start":"*","traggle_donw":"v","dash_dot":"-","dash_line":"--","x-mark":"x"}
    plt.figure(figsize = (12,8))
    for co,key in zip(colors,kind.keys()):
        y = np.random.randn(50)
        plt.plot(x,y,kind[key],color = co, label = key)
    plt.legend()
fun1()
```

```python
#标准直角坐标系下绘制三角函数
def fun2():
    x = np.linspace(-np.pi,np.pi,256)
    c = np.cos(x)
    s = np.sin(x) 
    plt.figure(figsize=(8,6))
    plt.plot(x, c, "-") 
    plt.plot(x, s, "--")
    
    #指定x轴和y轴的取值范围
    plt.xlim([-4,4]) 
    plt.ylim([-1,1])
    
    #修改刻度
    plt.xticks([-np.pi, -np.pi/2, 0, np.pi/2, np.pi],["$-\pi$","$-\pi/2$","$0$","$\pi/2$","$\pi$"]) 
    plt.yticks([-1,0,1])
    
    #去除上、右边框 
    ax = plt.gca()
    ax.spines["right"].set_color("none") 
    ax.spines["top"].set_color("none")
    #将底边框移到上面0点位置
    ax.spines["bottom"].set_position(("data",0))
    #将左边框移到右侧0点位置
    ax.spines["left"].set_position(("data",0))

    #在图上标点
    t = 2 * np.pi / 3
    y = np.sin(t)
    plt.scatter(t, y, 50, color="r")
    plt.plot((t,t),(0,y),"--", color="r")
    plt.plot((0,t),(y,y),"--", color="r")
    plt.annotate('($2\pi/3$,$\sqrt{3}/2$)',(t,y))
fun2()
```

```python
#纵向柱状图 
def fun3():
    price = [15000,23000,30000,50000]
    #第一个参数表示x的点，第二个参数表示y的点
    plt.bar(np.arange(4),price, align="center", color="steelblue", alpha=0.8,width = 0.5) 
    plt.xticks(np.arange(4),["Bj","SH","sz","WH"])
    plt.title("Price") 
fun3()
```

```python
#横向柱状图 
def fun4():
    price=[15000,23000,30000,50000]#第一个参数表示y的点，第二个参数表示×的点
    plt.barh(np.arange(4),price,align="center", color="steelblue") 
    plt.xlim([0,70000])
    for x,y in zip(np.arange(4),price):
        plt.text(y,x,y)
fun4()
```

```python
#画出两个柱状图的对比图
def fun1():
    x = np.arange(4)
    y2017 = [13500,14500, 25000,40000] 
    y2018 =[16000,24000,37000,55000]
    labels = ["北京","深圳","上海","杭州"]

    #设置每根柱子的宽度
    bar_width = 0.4
    plt.bar(x, y2017, width = bar_width, color = 'steelblue', alpha = 0.8)
    plt.bar(x + bar_width, y2018, width = bar_width, color = 'indianred', alpha = 0.8)

    #改变下标
    plt.xticks(x+0.2,labels)

    #添加文字
    for x1,y1 in zip(x,y2017):
        plt.text(x1-0.15,y1+500,y1)

    for x2,y2 in zip(x,y2018):
        plt.text(x2+0.25,y2+500,y2)
fun1()
```
